## Question 1 — LIKE

Find students whose names start with the letter `A`. Return `student_id`, `student_name`, and `email`.

## Question 2 — BETWEEN

Find attendance records whose `attendance_id` is between 10 and 25, inclusive. Return `attendance_id`, `student_key`, `class_key`, and `attendance_status`.

## Question 3 — Aggregation

Calculate the overall number of Present, Late, and Absent records using conditional aggregation. Return one row with `present_count`, `late_count`, and `absent_count`.

## Question 4 — Class Summary

Using `dim_class` and `fact_attendance`, calculate the number of attendance records for each class. Return `class_id`, `topic`, and `attendance_records`.

## Question 5 — Average

Using `fact_attendance`, calculate the average `attendance_count` for each `attendance_status`. Return `attendance_status` and `average_attendance_count`.

## Question 6 — LEFT JOIN + HAVING

Find students whose total number of attendance records is at least 2. Use `dim_student` and `fact_attendance`, and return `student_id`, `student_name`, and `attendance_records`.
